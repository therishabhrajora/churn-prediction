```python
import pandas as pd
import numpy as np
import joblib

best_model=joblib.load("../models/best_model.pkl")
def predict_churn(customer_data):

    customer_df = pd.DataFrame([customer_data])

    probability = best_model.predict_proba(
        customer_df
    )[0][1]

    prediction = int(
        best_model.predict(customer_df)[0]
    )

    if probability >= 0.70:
        risk = "HIGH"

    elif probability >= 0.40:
        risk = "MEDIUM"

    else:
        risk = "LOW"

    return {
        "prediction": (
            "Likely to Churn"
            if prediction == 1
            else "Likely to Stay"
        ),
        "churn_probability": round(
            probability * 100,
            2
        ),
        "risk_level": risk
    }

customer ={
    "gender": "Female",
    "SeniorCitizen": 1,
    "Partner": "No",
    "Dependents": "No",
    "tenure": 3,
    "PhoneService": "Yes",
    "MultipleLines": "Yes",
    "InternetService": "Fiber optic",
    "OnlineSecurity": "No",
    "OnlineBackup": "No",
    "DeviceProtection": "No",
    "TechSupport": "No",
    "StreamingTV": "Yes",
    "StreamingMovies": "Yes",
    "Contract": "Month-to-month",
    "PaperlessBilling": "Yes",
    "PaymentMethod": "Electronic check",
    "MonthlyCharges": 104.40,
   "TotalCharges": 313.20
}


result = predict_churn(customer)

print(result)
```
